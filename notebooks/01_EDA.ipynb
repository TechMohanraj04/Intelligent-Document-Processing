```python
# Import libraries

import os
import cv2
import matplotlib.pyplot as plt
import pandas as pd
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
invoice_path = "/content/drive/MyDrive/data/invoices/train/images"

resume_path = "/content/drive/MyDrive/data/resumes"

id_path = "/content/drive/MyDrive/data/id_cards/Aadharcard/train/images"
```

```python
invoice_count = len(os.listdir(invoice_path))

resume_count = len(os.listdir(resume_path))

id_count = len(os.listdir(id_path))

print("Invoices :", invoice_count)

print("Resumes :", resume_count)

print("ID Cards :", id_count)
```

```python
# Display sample invoice

sample = os.listdir(invoice_path)[0]

img = cv2.imread(os.path.join(invoice_path, sample))

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img)

plt.axis("off")
```

```python
# Display 5 invoices

fig = plt.figure(figsize=(15,8))

for i,file in enumerate(os.listdir(invoice_path)[:5]):

    img = cv2.imread(os.path.join(invoice_path,file))

    img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

    plt.subplot(1,5,i+1)

    plt.imshow(img)

    plt.axis("off")
```

```python
for file in os.listdir(invoice_path)[:5]:

    img = cv2.imread(os.path.join(invoice_path,file))

    print(file,img.shape)
```

```python
extensions={}

for file in os.listdir(invoice_path):

    ext=file.split(".")[-1]

    extensions[ext]=extensions.get(ext,0)+1

print(extensions)
```

```python
# Creating a DataFrame

summary=pd.DataFrame({

"Dataset":["Invoices","Resumes","ID Cards"],

"Count":[invoice_count,resume_count,id_count]

})

summary
```

```python
# Plot dataset distribution

summary.plot.bar(x="Dataset",y="Count")
```

```python
# Image Width Distribution

widths = []

for file in os.listdir(invoice_path):

    img = cv2.imread(os.path.join(invoice_path,file))

    widths.append(img.shape[1])
```

```python
plt.hist(widths,bins=20)

plt.title("Invoice Image Width")

plt.xlabel("Pixels")

plt.ylabel("Count")

plt.show()
```

```python
fig=plt.figure(figsize=(15,8))

files=os.listdir(invoice_path)[:6]

for i,file in enumerate(files):

    img=cv2.imread(os.path.join(invoice_path,file))

    img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

    plt.subplot(2,3,i+1)

    plt.imshow(img)

    plt.axis("off")

plt.tight_layout()

plt.show()
```
